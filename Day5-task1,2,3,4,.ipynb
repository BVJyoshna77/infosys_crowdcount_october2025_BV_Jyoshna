```python
#task1

#fiboonaci of a number using recurrsive functions
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
fib_input = 6
print("Fibonacci of", fib_input, "is", fibonacci(fib_input))
```

```python
#task1

#factorial of a number using recurrsive fuctions
def factorial(n):
    if n == 0:
        return 1  # Base case: 0! = 1
    else:
        return n * factorial(n - 1)  # Recursive case
print(factorial(5)) 
```
