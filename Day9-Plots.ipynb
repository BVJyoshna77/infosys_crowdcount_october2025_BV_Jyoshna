```python
import pandas as pd
df = pd.DataFrame({'w': [1, 2, 3, 4, 5]})
df['w'].hist()
```

```python
df.plot(y='w', kind='line', title='Line Plot of w', marker='o')
#line plot
```

```python
import pandas as pd
df = pd.DataFrame({
    'w': [1, 2, 3, 4],
    'x': [5, 6, 7, 8]
})
```

```python
df[['w', 'x']].plot(kind='bar', title='Bar Plot of w and x', color=['red', 'blue'])
```

```python
df[['w', 'x']].plot(kind='bar', stacked=True, title='Stacked Bar Plot')
#stacked bar plot
```

```python
df.plot(kind='box', title='Box Plot of All Columns')
#box plot
```

```python
df.plot(kind='scatter', x='w', y='x', color='purple')
#scatter plot
```

```python
df['w'].value_counts().plot(kind='pie', autopct='%2.1f%%')
#pie chart
```
