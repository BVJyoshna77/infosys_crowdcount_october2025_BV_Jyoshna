```python
 #Data Cleaning
# Load the dataset
df = pd.read_csv(r'C:\Users\jyosh\Downloads\archive\Coffe_sales.csv')

# 1. Check missing values
print(df.isna().sum())

# 2. Remove duplicates
df = df.drop_duplicates()

# 3. Convert 'Date' to datetime type
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# 4. Strip whitespace from string columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

# 5. Optional: Correct other columns' types if needed (ints, floats)
# e.g., df['hour_of_day'] = df['hour_of_day'].astype(int)

# Save the cleaned data
df.to_csv('Coffe_sales_cleaned.csv', index=False)

print("Data cleaning complete. Cleaned file saved as 'Coffe_sales_cleaned.csv'")
```
