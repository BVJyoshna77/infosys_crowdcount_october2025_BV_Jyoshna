```python
#task1 program on calculator using if elif else
num1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /, %, //, **): ")
num2 = float(input("Enter the second number: "))

if operator == '+':
    result = num1 + num2
elif operator == '-':
    result = num1 - num2
elif operator == '*':
    result = num1 * num2
elif operator == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero"
elif operator == '%':
    if num2 != 0:
        result = num1 % num2
    else:
        result = "Error: Division by zero"
elif operator == '//':
    if num2 != 0:
        result = num1 // num2
    else:
        result = "Error: Division by zero"
elif operator == '**':
    result = num1 ** num2
else:
    result = "Error: Invalid operator"
print(f"Result: {result}")
```

```python
#task2 for loop using tuple,dictionary,set

#using tuple
my_tuple = (10, 20, 30, 40)
for item in my_tuple:
    print(f"Tuple item: {item}")

# For loop with a dictionary 
my_dict = {'a': 1, 'b': 2, 'c': 3}
for key in my_dict:
    print(f"Key: {key}, Value: {my_dict[key]}")

# Alternatively, iterating key-value pairs
for key, value in my_dict.items():
    print(f"Key: {key}, Value: {value}")

# For loop with a set
my_set = {100, 200, 300}
for element in my_set:
    print(f"Set element: {element}")
```

```python
#task3 design calculator using structuralpattern matching
def calculate(operation):
    match operation:
        case ("+", x, y):
            return x + y
        case ("-", x, y):
            return x - y
        case ("*", x, y):
            return x * y
        case ("/", x, y):
            if y == 0:
                return "Error: Division by zero"
            return x / y
        case _:
            return "Error: Unsupported operation"
result = calculate(("+", 5, 3))
print(result)
```

```python
#task 4 writing all file concepts
import os

# Writing to a file ('w' mode)
with open('myfile.txt', 'w') as f:
    f.write('Hello World\nThis is a sample file.\n')

# Reading the entire file
with open('myfile.txt', 'r') as f:
    print('--- Full Content ---\n', f.read())

# Reading line by line
with open('myfile.txt', 'r') as f:
    print('--- Line by Line ---')
    for line in f:
        print(line.strip())

#appending 
with open('myfile.txt', 'a') as f:
    f.write('Appended line is here.\n')
#seek and tell operations
with open('myfile.txt', 'r') as f:
    f.seek(6) 
    print('After seek:', f.read())
    print('Current pointer:', f.tell())

# Exception handling 
try:
    with open('no_file.txt', 'r') as nf:
        print(nf.read())
except FileNotFoundError:
    print('FileNotFoundError handled!')

# Binary file example:
with open('binfile.dat', 'wb') as bf:
    bf.write(b'\x00\x01\x02')
with open('binfile.dat', 'rb') as bf:
    print('Binary file content:', bf.read())

# Exclusive creation - 'x' mode
try:
    with open('unique.txt', 'x') as uf:
        uf.write('Created with x mode.\n')
except FileExistsError:
    print('unique.txt already exists.')

# Delete a file
if os.path.exists('unique.txt'):
    os.remove('unique.txt')
    print('unique.txt deleted.')

#List all files in the current directory
print('Files in the directory:')
for file in os.listdir('.'):
    if os.path.isfile(file):
        print(file)
```
