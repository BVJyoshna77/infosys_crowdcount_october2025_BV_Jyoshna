```python
#task 1 google form with all 5 methods
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height in meters: "))
is_student = input("Are you a student : ") == 'True'
hobbies = input("Enter your hobbies as comma-separated values : ").split(',')
info = eval(input("Enter your info as a dictionary : "))
```

```python
# 1. Using comma
print("Name:", name, "Age:", age, "Height:", height, "Student:", is_student, "Hobbies:", hobbies, "Info:", info)
```

```python
# 2. Using plus (+)
print("Name: " + name + " Age: " + str(age) + " Height: " + str(height) + " Student: " + str(is_student) + 
      " Hobbies: " + str(hobbies) + " Info: " + str(info))
```

```python
# 3. Using percent (%)
print("Name: %s Age: %d Height: %.2f Student: %s Hobbies: %s Info: %s" %
      (name, age, height, is_student, hobbies, info))
```

```python
# 4. Using .format()
print("Name: {} Age: {} Height: {} Student: {} Hobbies: {} Info: {}".format(
    name, age, height, is_student, hobbies, info))
```

```python
# 5. Using f-string
print(f"Name: {name} Age: {age} Height: {height} Student: {is_student} Hobbies: {hobbies} Info: {info}")
```
